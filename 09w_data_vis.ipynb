```python
import sys
if 'pydataset' not in sys.modules:
    !pip install pydataset
from pydataset import data

# 기니피그의 치아 성장(ToothGrowth) 데이터를 가져옵니다.
tg = data('ToothGrowth')

# 데이터프레임의 요약 정보를 출력합니다.
tg.info()
```

```python
plt.scatter(tg.dose, tg.len)
plt.title("기니피그의 비타민과 치아성장 관계")
plt.xlabel("비타민 용량")
plt.ylabel("치아성장 세포길이")
plt.show()
```

```python
import seaborn as sns
sns.scatterplot(data = tg, x = 'dose', y = 'len', hue = 'supp')
plt.title("기니피그의 비타민과 치아성장 관계")
plt.xlabel("비타민 용량")
plt.ylabel("치아성장 세포길이")
plt.show()
```

```python
# 시각화
sns.boxplot(data=tg, x='dose', y='len', hue='supp')
sns.stripplot(data=tg, x='dose', y='len', hue='supp',
dodge=True, alpha=0.6)

plt.title('Tooth Growth by Dose and Supplement Type')
plt.ylabel('Tooth Length (mm)')
plt.xlabel('Dose (mg/day)')
plt.legend(title='Supplement')
plt.tight_layout()
plt.show()
```

```python
# 시각화
sns.boxplot(data=tg, x='dose', y='len', hue='supp')
sns.stripplot(data=tg, x='dose', y='len', hue='supp',
dodge=True, alpha = .4)

plt.title('기니피그 치아 성장')
plt.ylabel('치아길이 (mm)')
plt.xlabel('용량(mg/day)')
plt.legend(title='Supplement', loc = 'right')
plt.tight_layout()
plt.show()
```

```python
sns.countplot(mtc, x = 'gear');
```

```python
ax= sns.countplot(data = mtc, x='gear',
                   order = mtc.gear.value_counts().index)
for  p in ax.patches:
    height = p.get_height()
    ax.text(x = (p.get_x() + p.get_width() / 2.), y= (height +.2),
            s = int(height), ha = 'center', size = 12, color = 'blue')
    ax.set_ylim(0, 17)
    plt.show()
```
